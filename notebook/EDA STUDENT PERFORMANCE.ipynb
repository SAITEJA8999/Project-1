```python
import matplotlib.pyplot as plt

# 1. Ensure all columns are lowercase and stripped
df.columns = df.columns.str.strip().str.lower()

# 2. Set figure parameters
plt.rcParams['figure.figsize'] = (30, 12)

# Subplot 1: Gender
plt.subplot(1, 5, 1)
size = df['gender'].value_counts()
color = ['red', 'green']
plt.pie(size, colors=color, labels=size.index, autopct='%.2f%%') # Safe dynamic labels & fixed autopct
plt.title('Gender', fontsize=20)
plt.axis('off')

# Subplot 2: Race/Ethnicity (Fixed column name from 'race/ethnicity' to 'race_ethnicity')
plt.subplot(1, 5, 2)
size = df['race_ethnicity'].value_counts()
color = ['red', 'green', 'blue', 'cyan', 'orange']
plt.pie(size, colors=color, labels=size.index, autopct='%.2f%%')
plt.title('Race/Ethnicity', fontsize=20)
plt.axis('off')

# Subplot 3: Lunch
plt.subplot(1, 5, 3)
size = df['lunch'].value_counts()
color = ['red', 'green']
plt.pie(size, colors=color, labels=size.index, autopct='%.2f%%')
plt.title('Lunch', fontsize=20)
plt.axis('off')

# Subplot 4: Test Course (Fixed column name from 'test preparation course' to 'test_preparation_course')
plt.subplot(1, 5, 4)
size = df['test_preparation_course'].value_counts()
color = ['red', 'green']
plt.pie(size, colors=color, labels=size.index, autopct='%.2f%%')
plt.title('Test Course', fontsize=20)
plt.axis('off')

# Subplot 5: Parental Education (Fixed column name from 'parental level of education' to 'parental_level_of_education')
plt.subplot(1, 5, 5)
size = df['parental_level_of_education'].value_counts()
color = ['red', 'green', 'blue', 'cyan', 'orange', 'grey']
plt.pie(size, colors=color, labels=size.index, autopct='%.2f%%')
plt.title('Parental Education', fontsize=20)
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 8))

X = ['Total Average', 'Math Average']

# Use .loc with explicit gender names and fix 'math_score' column name
female_scores = [gender_group.loc['female', 'average'], gender_group.loc['female', 'math_score']]
male_scores = [gender_group.loc['male', 'average'], gender_group.loc['male', 'math_score']]

X_axis = np.arange(len(X))

# Kept bars aligned with labels (Male on the left (-0.2), Female on the right (+0.2))
plt.bar(X_axis - 0.2, male_scores, 0.4, label='Male', color='green')
plt.bar(X_axis + 0.2, female_scores, 0.4, label='Female', color='red')

plt.xticks(X_axis, X)
plt.ylabel("Marks")
plt.title("Total average v/s Math average marks of both the genders", fontweight='bold')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Group by the correct column name
Group_data2 = df.groupby('race_ethnicity')

# 2. Setup the 1-row, 3-column figure layout
f, ax = plt.subplots(1, 3, figsize=(20, 8))

# Subplot 1: Math Score (Fixed column names and added numeric_only=True)
math_means = Group_data2['math_score'].mean(numeric_only=True)
sns.barplot(x=math_means.index, y=math_means.values, palette='mako', ax=ax[0])
ax[0].set_title('Math score', color='#005ce6', size=20)
for container in ax[0].containers:
    ax[0].bar_label(container, color='black', size=15)

# Subplot 2: Reading Score (Fixed column names and added numeric_only=True)
reading_means = Group_data2['reading_score'].mean(numeric_only=True)
sns.barplot(x=reading_means.index, y=reading_means.values, palette='flare', ax=ax[1])
ax[1].set_title('Reading score', color='#005ce6', size=20)
for container in ax[1].containers:
    ax[1].bar_label(container, color='black', size=15)

# Subplot 3: Writing Score (Fixed column names and added numeric_only=True)
writing_means = Group_data2['writing_score'].mean(numeric_only=True)
sns.barplot(x=writing_means.index, y=writing_means.values, palette='coolwarm', ax=ax[2])
ax[2].set_title('Writing score', color='#005ce6', size=20)
for container in ax[2].containers:
    ax[2].bar_label(container, color='black', size=15)

plt.tight_layout()
plt.show()
```
